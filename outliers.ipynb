```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("C:/Users/user/Downloads/Titanic-Dataset.csv")
df
```

```python
num_cols = ['Age', 'Fare', 'SibSp', 'Parch']
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nOutliers in {col}:")
    print(outliers[[col]])

    print("Number of outliers:", len(outliers))
```

```python
for col in num_cols:
    mean = df[col].mean()
    std = df[col].std()

    z_score = (df[col] - mean) / std

    outliers = df[abs(z_score) > 3]

    print(f"\nOutliers in {col}:")
    print(outliers[[col]])

    print("Number of outliers:", len(outliers))
```

```python
df_capped = df.copy()

for col in num_cols:
    Q1 = df_capped[col].quantile(0.25)
    Q3 = df_capped[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_capped[col] = df_capped[col].clip(lower, upper)

print("Outliers capped successfully!")
```

```python
df_z_capped = df.copy()

for col in num_cols:
    mean = df_z_capped[col].mean()
    std = df_z_capped[col].std()

    lower = mean - 3 * std
    upper = mean + 3 * std

    df_z_capped[col] = df_z_capped[col].clip(lower, upper)

print("Z-score capping completed!")
```
