```python
data=pd.read_csv("C:/Users/user/Downloads/Titanic-Dataset.csv")
```

```python
df=pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df.isnull().sum()
```

```python
df['Survived']=df['Survived'].fillna(df['Survived'].mean())
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df['Embarked'] = df['Embarked'].ffill()
```

```python
df.isnull().sum()
```

```python
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['Sex'] = df['Sex'].map({'male': 0,'female': 1})
```

```python
print(df[['Name', 'Sex']].head())
```

```python
df = pd.get_dummies(
    df,
    columns=['Embarked', 'Pclass'],
    dtype=int
)
```

```python
df = df.drop(
    columns=['Cabin'],
    errors='ignore'
)
```
