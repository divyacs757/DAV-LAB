```python
import pandas as pd
from scipy import stats
import numpy as np

df = {
    'name': ['divya', 'priyanka', 'chaitra', 'bhoomika', 'sanjana',
             'dhanu', 'guru', 'darshan', 'gani', 'lokesh'],

    'marks': [80, 85, 45, 60, 79, 80, 98, 45, 78, 90],

    'branch': ['cse', 'aiml', 'cse', 'aiml', 'cse',
               'cse', 'aiml', 'cse', 'aiml', 'cse']
}

df = pd.DataFrame(df)

print(df)
```

```python
print("Mean:", df['marks'].mean())
print("Median:", df['marks'].median())
print("Mode:", df['marks'].mode()[0])
```

```python
print("Variance:", df['marks'].var())
print("Standard Deviation:", df['marks'].std())
```
