```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("D:/dip(AIML(cse))/kaggle/studentsinfo.csv")
```

```python
plt.hist(df["marks_scince"].dropna(), bins=5)

plt.xlabel("Science Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Science Marks")

plt.show()
```

```python
plt.boxplot([
    df["marks_math"].dropna(),
    df["marks_scince"].dropna()
])

plt.xticks([1, 2], ["Math", "Science"])
plt.ylabel("Marks")
plt.title("Math vs Science Marks")

plt.show()
```

```python
plt.scatter(
    df["study_hours"],
    df["marks_math"]
)

plt.xlabel("Study Hours")
plt.ylabel("Math Marks")
plt.title("Study Hours vs Math Marks")

plt.show()
```

```python
numeric_data = df[
    [
        "age",
        "study_hours",
        "attendance(%)",
        "marks_math",
        "marks_scince"
    ]
]

correlation = numeric_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```
